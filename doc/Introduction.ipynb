# Topics: problems to solve

A data-driven application needs to handle diverse, fast-changing **data** into **information** people can use safely and reliably.

:::: {.columns}
::: {.column width="50%" .fragment}

1. **Acquire and integrate**: Bring together data from different sources, formats, and APIs—by batch or as events arrive.
2. **Move and process**: Validate, transform, combine, and analyze data at the scale and latency the use case needs.

:::
::: {.column width="50%" .fragment}

3. **Store and manage**: Keep data durable and discoverable as schemas and requirements evolve; consider quality, access, and lineage.
4. **Find and use**: Support analytics, visualization, search, machine learning, and AI applications with relevant, trustworthy data.
:::
::::

:::{.callout-tip .fragment}
What data is needed, how quickly, at what scale, and with which quality, privacy, reliability, and cost constraints?
:::

## From problems to an evolving ecosystem

The four problems on the previous slide (acquire, move, store, find) haven't changed, but *how* systems solve them has evolved as data volume, velocity, and variety grew.

:::: {.columns}
::: {.fragment .column width="50%"}
**Two levels**

- **Architecture patterns**: how acquisition, processing, and storage fit together, and how that fit has changed over time.
- **Per-topic technologies**: real-world tools for each stage of the pipeline.
:::
::: {.fragment .column width="50%"}
**The story we'll trace**

1. Batch vs. streaming: a trade-off between freshness and simplicity.
2. Lambda architecture: an early fix that runs both, at a cost.
3. Lakehouse: a more recent fix that removes that cost.
:::

## Batch vs. streaming processing

Two complementary ways to move and process data, differing in when work happens and how fresh the results are.

:::: {.columns}
::: {.fragment .column width="50%"}
**Batch processing**

- Process bounded, already-collected data on a schedule (hourly, nightly, ...).
- *You gain:* high throughput, simple to reason about, easy to retry/reprocess from scratch.
- *You pay:* latency: results reflect data as of the last run, not right now. (and also costs)
- Example: nightly re-index of course documents, data warehouse ETL, or a nightly report of the day's sales.
:::
::: {.fragment .column width="50%"}
**Stream processing**

- Process unbounded events as they arrive, one at a time or in small windows.
- *You gain:* low latency: results available continuously, close to real time.
- *You pay:* complexity:ordering, late data, state, and exactly-once semantics.
- Example: an incoming Telegram message triggering the assistant.
:::
::::

:::{.callout-tip .fragment}
Does the use case need an answer now, or is a periodic refresh good enough? Many real systems need *both* ?
:::

## The Lambda architecture (2011)

::: {.columns}
::: {.fragment .column width="50%"}
**The problem:** early big-data systems needed both a correct, complete historical view *and* a low-latency view of recent events—batch alone was too slow, streaming alone was hard to make fully correct.
::: 
::: {.fragment .column width="50%"}
**The fix** (proposed by Nathan Marz creator of Apache Storm): run batch and streaming *side by side*, over the same raw data, and merge their results.
::: 
:::: 
:::: {.columns}
::: {.fragment .column width="50%"}
**Three layers**

- **Batch layer:** recomputes precise, complete views over all historical data (e.g., with Spark/Hadoop).
- **Speed layer:** processes recent events immediately (e.g., with Kafka Streams), approximating the batch view until it catches up.
- **Serving layer:** merges batch and speed views to answer queries.
:::
::: {.fragment .column width="50%"}
**Trade-off**

- *You gain:* correctness (batch) plus freshness (speed) for the same data.
- *You pay:* two codebases/pipelines implementing similar logic twice, which must be built, tested, and kept consistent.
- This cost is the reason it was later challenged and alternatives emerged.
:::
::::

:::{.callout-note .fragment}
 Jay Kreps (creator of Apache Kafka and co-founder of Confluent) in ["Questioning the Lambda Architecture"](https://www.oreilly.com/radar/questioning-the-lambda-architecture/) (O'Reilly Radar, 2014) proposes the single-pipeline **Kappa architecture** as a response
:::

## Evolution toward Lakehouse architectures

Jay Kreps' critique motivated simplifying Lambda's two pipelines into one. Storage evolved the same way, to remove duplication at the data layer too.

:::: {.columns}
::: {.fragment .column width="50%"}

Data warehouse:
 : Structured, curated data; strong schema and query performance.
 : Costly to scale, and not suited to raw/unstructured or ML data.

Data lake:
 : Raw files at scale, any format, low storage cost (e.g., HDFS, MinIO, RustFS).
 : Weak consistency and governance: easy to turn into a "data swamp".
:::
::: {.fragment .column width="50%"}

Lakehouse: 
 : Table formats (Apache Iceberg, Delta Lake) add ACID transactions, schema evolution, and time travel on top of lake storage.
 : The same table can be read by a batch job and written by a streaming job—one copy of the data, not two pipelines.
 : Solves the *storage* side of the Lambda trade-off—but storage is only one stage of the pipeline.
:::
::::   

:::{.callout-tip .fragment}
The bigger picture:** acquiring, moving, searching, and using data each still raise their own design questions
:::

# Technologies:  tools to fix the problem


## Data Ingestion

Bring data from applications, logs, and APIs into the pipeline, by batch or as events arrive.

:::: {.columns}
::: {.fragment .column width="50%"}
**Technologies**

- Apache Flume
- Apache NiFi
- Logstash
- Fluentd
- n8n
:::
::: {.fragment .column width="50%"}
**Applications**

- Collect and reliably move events from application logs into storage.
- Route, transform, and monitor data flows across heterogeneous systems.
- Parse, enrich, and route application and infrastructure logs.
- Aggregate logs from distributed services and forward them to a searchable store.
- Connect APIs and automate operational workflows with low-code orchestration.
:::
::::

## Data Streaming

Buffer, distribute, and process events in real time, connecting producers and consumers asynchronously.

:::: {.columns}
::: {.fragment .column width="50%"}
**Technologies**

- Apache Kafka (Connect, Streams, Schema Registry)
- Confluent
- Redpanda
- Apache Pulsar
- NATS
:::
::: {.fragment .column width="50%"}
**Applications**

- Buffer and distribute events for decoupled services and real-time pipelines.
- Move records between Kafka and external systems without custom producer/consumer code.
- Implement lightweight, stateful stream transformations directly on Kafka topics.
- Enforce and evolve message schemas; a data dictionary for event streams.
- Operate Kafka-based pipelines with managed connectors and platform tooling.
:::
::::

## Data Processing

Transform, validate, and analyze data at the scale and latency the use case needs, in batch or streaming.

:::: {.columns}
::: {.fragment .column width="50%"}
**Technologies**

- Apache Spark (with Spark NLP)
- Apache Flink
- Ray
- DuckDB
:::
::: {.fragment .column width="50%"}
**Applications**

- Transform large datasets in batch and streaming workloads; process Lakehouse tables with Iceberg.
- Process and analyze text at scale within Spark pipelines, including LLM-assisted tasks.
- Alternative to Spark; process unbounded streams with low latency and exactly-once semantics.
- Alternative to Spark/Flink; scale Python workloads, including ML model training and inference.
- Alternative to Spark; run SQL queries on local or in-memory datasets, including Parquet and Arrow files.
:::
::::
