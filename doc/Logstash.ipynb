### Filters

::::{.columns}

::: {.fragment .column width="50%"}
**Parse & transform your data on the fly**

As data travels from source to store, Logstash filters parse each event, identify named fields to build structure, and transform them to converge on a common format for more powerful analysis and business value.

Logstash dynamically transforms and prepares your data regardless of format or complexity:

* Derive structure from unstructured data with grok
* Decipher geo coordinates from IP addresses
* Anonymize PII data, exclude sensitive fields completely
* Ease overall processing, independent of the data source, format, or schema.

## Features

::::{.columns}

::: {.fragment .column width="50%"}
**Extensibility**

- Logstash has a pluggable framework featuring over 200 plugins. Mix, match, and orchestrate different inputs, filters, and outputs to work in pipeline harmony.
- Build your own plugin using Ruby and package using gem
- See [reference](https://www.elastic.co/guide/en/logstash/current/input-new-plugin.html)
:::

::: {.fragment .column width="50%"}
**Durability & Security** 

- Logstash guarantees at-least-once delivery for your in-flight events with its persistent queue.
- Events that are not successfully processed can be shunted to a dead letter queue for introspection and replay.
- Ability to absorb throughput, Logstash scales through ingestion spikes without having to use an external queueing layer.
- Fully secure your ingest pipelines.

# More about Logstash

## Inside an event: parsing and schema

A Logstash event is a structure of fields. Common fields include `@timestamp`, `@version`, `message`, and `tags`; plugins and filters can add nested fields.

```json
{
  "@timestamp": "2026-10-10T10:15:30Z",
  "message": "203.0.113.8 GET /orders 500",
  "client": { "ip": "203.0.113.8" },
  "http": { "response": { "status_code": 500 } },
  "tags": []
}
```

- **Dissect**: stable delimiters and lightweight parsing; **Grok**: variable formats and regex-based patterns.
- **Date**: convert the source timestamp into `@timestamp`; otherwise, analysis may use the ingestion time instead of the event time.
- **ECS**: a shared schema (`client.ip`, `http.response.status_code`, …) that helps keep queries and dashboards consistent across sources.

Sources: [events and field references](https://www.elastic.co/docs/reference/logstash/event-dependent-configuration) · [Grok](https://www.elastic.co/docs/reference/logstash/plugins/plugins-filters-grok) · [Dissect](https://www.elastic.co/docs/reference/logstash/plugins/plugins-filters-dissect) · [Date filter](https://www.elastic.co/docs/reference/logstash/plugins/plugins-filters-date) · [Elastic Common Schema](https://www.elastic.co/docs/reference/ecs)

## Execution and performance: measure before tuning

- Each pipeline processes events in **batches**; workers execute filters and outputs. Larger batches can reduce overhead, but increase memory use and latency.
- More `pipeline.workers` can help plugins waiting on network or I/O; it does not automatically speed up CPU-bound filters and may increase contention.
- Throughput depends on the bottleneck: source, parsing, CPU, disk, network, or destination. A slow Elasticsearch destination can slow the upstream pipeline.
- Monitor `worker_utilization`, `worker_millis_per_event`, and `queue_backpressure`; change one setting at a time and compare results.
- Event ordering is not guaranteed by default. If needed, a pipeline can preserve order with a single worker and `pipeline.ordered: true`, at a performance cost.

Sources: [tuning and flow metrics](https://www.elastic.co/docs/reference/logstash/tuning-logstash) · [processing and event ordering](https://www.elastic.co/docs/reference/logstash/processing)

## Reliability: Persistent Queues and Dead Letter Queues

- **Persistent Queue (PQ):** an on-disk buffer, disabled by default; it helps recover in-flight events after a restart and absorb bursts. It is limited by disk capacity and applies backpressure when full.
- **At-least-once** delivery depends on inputs with acknowledgements and on checkpoints; not all inputs are protected. It is not exactly-once: events may be duplicated during recovery, so design outputs and consumers to be idempotent.
- A PQ is not a replicated backup: it does not protect against permanent disk or machine loss, and protection also depends on the input protocol and checkpoints.
- **Dead Letter Queue (DLQ):** stores some events that cannot be processed, such as specific Elasticsearch mapping errors. It is not a general-purpose buffer: network errors and retries are not automatically routed to the DLQ.

Sources: [Persistent Queues: benefits and limitations](https://www.elastic.co/docs/reference/logstash/persistent-queues) · [Dead Letter Queues: supported cases](https://www.elastic.co/docs/reference/logstash/dead-letter-queues)

## From demo to production

- **Availability:** run multiple Logstash nodes behind a load balancer; for inputs with acknowledgements, use persistent queues and also ensure storage redundancy.
- **Multiple flows:** `pipelines.yml` enables separate pipelines, each with independent configuration and resource settings; they still compete for CPU, memory, and disk.
- **Observability:** the Monitoring API exposes node, JVM, pipeline, and plugin statistics (on port `9600` by default). Restrict access and enable authentication/TLS as appropriate for your environment.
- **Data security:** encrypt connections, use least-privilege credentials, and store them in the Logstash Keystore or environment variables; limit personal data collection and retention.

Sources: [deployment and high availability](https://www.elastic.co/docs/reference/logstash/deploying-scaling-logstash) · [multiple pipelines](https://www.elastic.co/docs/reference/logstash/multiple-pipelines) · [Monitoring API and security](https://www.elastic.co/docs/reference/logstash/monitoring-logstash) · [secure connections to Elasticsearch](https://www.elastic.co/docs/reference/logstash/secure-connection)
