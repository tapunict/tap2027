## Life of a Fluentd event
<https://docs.fluentd.org/quickstart/life-of-a-fluentd-event>

:::: {.columns}

::: {.column width="50%" .fragment}
**Event** = `tag` (origin, used for routing) + `time` (nanosecond resolution) + `record` (JSON log)

1. **Source** (input plugin) generates the event, e.g. `in_http` or `in_tail`
2. **Filter**s accept or reject/modify it, in order from top to bottom
3. **Match** routes it by tag to an **Output** plugin

- **Labels** (`@label`) define routing sections that are linked references instead of top-to-bottom order, keeping complex configs readable
- **Buffers**: stdout for development, sends immediately, in production buffered outputs (`forward`, `s3`, ...) events are stored, then flushed to the destination 
:::

::: {.column width="50%" .fragment}
```config
<source>
  @type http
  port 8888
  @label @STAGING
</source>

<label @STAGING>
  <filter test.cycle>
    @type grep
    <exclude>
      key action
      pattern ^logout$
    </exclude>
  </filter>
  <match test.cycle>
    @type stdout
  </match>
</label>
```
`curl -X POST -d 'json={"action":"login","user":2}' http://localhost:8888/test.cycle` is printed; with `"logout"` it is discarded.
:::
::::

Source: [Life of a Fluentd event](https://docs.fluentd.org/quickstart/life-of-a-fluentd-event)

## Plugin types{background-color="black" background-image="https://images.unsplash.com/photo-1687038367723-4494d8e420f0?q=80&w=3000&auto=format&fit=crop&ixlib=rb-4.1.0&ixid=NHwxMjA3fDA%3D" background-size="cover" background-opacity="0.5" background-position="center"}
<https://docs.fluentd.org/input>

:::: {.columns}

::: {.fragment .column width="50%"}
Fluentd has 9 types of plugins:

- [Input](https://docs.fluentd.org/input): retrieve and pull event logs from the external sources
- [Parser](https://docs.fluentd.org/parser): depends on Input type, verify if the event input is syntactly correct. 
- [Filter](https://docs.fluentd.org/filter): allows to filter data, add/delete/mask fields
- [Output](https://docs.fluentd.org/output): sends data to destination
- [Formatter](https://docs.fluentd.org/formatter): For an output plugin that supports Formatter, the <format> directive can be used to change the output format.



::: 

::: {.fragment .column width="50%"}
- [Storage](https://docs.fluentd.org/storage): maintains durable information in the ingestion phase
- [Service Discovery](https://docs.fluentd.org/service_discovery): For output plugins, allows to define rules to select destinations (load balancing,...)
- [Buffer](https://docs.fluentd.org/buffer): Define chunks for output plugins
- [Metrics](https://docs.fluentd.org/metrics): To monitor fluentd
:::

## Fluent Bit: what it is{background-color="black" background-image="images/fluentbit.avif" background-size="100%" background-opacity="0.5" background-position="center"}

*"A super fast, lightweight, and highly scalable logging, metrics, and traces processor and forwarder. The preferred choice for cloud and containerized environments."* ([fluentbit.io](https://fluentbit.io/))

- **Efficient**: ~450 KB minimal footprint, zero external dependencies, asynchronous and event-driven design
- **Scalable**: >1 PB of data per day across thousands of sources and destinations; >1 billion sources managed (IoT devices, Windows and Linux servers)
- **Flexible**: parsing, filtering, enrichment and dynamic routing to multiple destinations (zero-copy)
- **Extensible**: 100+ built-in plugins for inputs, filters and outputs
- **OpenTelemetry-native**: OTLP ingestion and delivery
- "Deployed over fifteen billion times"

 

## Fluent Bit vs Fluentd
[from fluentbit.io](https://fluentbit.io/fluent-bit-vs-fluentd/)

| Attribute | Fluentd | Fluent Bit |
|---|---|---|
| Scope | Containers / Servers | Embedded Linux / Containers / Servers |
| Language | C and Ruby | C |
| Memory | > 60 MB | ~ 450 KB |
| Performance | Medium | High |
| Dependencies | Ruby gem, depends on other gems | Zero, unless required by a plugin |
| Plugins | 1,000+ external | 100+ built-in |
| OpenTelemetry | Via plugins | Native OTLP |
| License | Apache 2.0 | Apache 2.0 |


## FluentBit is the new Fluentd

- 2015 (The IoT Origin): Created by Eduardo Silva at Treasure Data as a pure-C, ultra-lightweight edge forwarder to complement the heavier, Ruby-based Fluentd.
- 2018 (The Architectural Shift): As Kubernetes clusters scaled, Fluentd's high memory/CPU footprint became a bottleneck; engineers shifted to Fluent Bit to drastically cut infrastructure costs.
- 2020 (The Kubernetes Takeover): Graduating within the CNCF, Fluent Bit effectively replaced Fluentd as the de facto standard node agent (DaemonSet) for containerized environments.
- 2022 (The Telemetry Evolution): It expanded far beyond simple log routing, evolving into a unified, vendor-neutral pipeline natively supporting OpenTelemetry Logs, Metrics, and Traces.
- 2024 (The Enterprise Standard): The original creators launched Calyptia (then Chronosphere, now Cortex XCOR leader in Observability) to manage massive enterprise fleets, cementing the reality that Fluent Bit is now the definitive "Fluentd".

## Example: observing a Telegram agent

The agent writes one JSON line per event. Fluent Bit keeps only the events that need attention, **redacts the user name** and tags the service.

:::: {.columns}

::: {.column width="50%"}
`agent.log` (simulated)
```json
{"ts":"…10:00:01Z","event":"message_received","chat_id":"c-1042","user":"mario.rossi"}
{"ts":"…10:00:02Z","event":"retrieval","chat_id":"c-1042","sources_count":3,"latency_ms":120}
{"ts":"…10:00:03Z","event":"answer_sent","chat_id":"c-1042","latency_ms":850}
{"ts":"…10:01:11Z","event":"declined","chat_id":"c-2210","user":"anna.verdi","reason":"no_supported_source"}
{"ts":"…10:02:35Z","event":"error","chat_id":"c-3307","user":"luca.bianchi","error":"elasticsearch timeout"}
{"ts":"…10:03:00Z","event":"handoff","chat_id":"c-4120","user":"sara.neri","reason":"low_confidence"}
```
Event types: `message_received`, `retrieval`, `answer_sent`, `declined`, `handoff`, `error`.
:::

::: {.column width="50%"}
`fluent-bit.yaml`
```yaml
service:
  parsers_file: /fluent-bit/etc/parsers.conf
pipeline:
  inputs:
    - name: tail
      path: /data/agent.log
      tag: agent.telegram
      parser: json
      read_from_head: true
      exit_on_eof: true
  filters:
    - name: grep
      match: agent.*
      regex: event ^(error|declined|handoff)$
    - name: record_modifier
      match: agent.*
      record: service tap-agent
      remove_key: user
  outputs:
    - name: stdout
      match: "*"
      format: json_lines
```
:::
::::

## Example: running it

```bash
docker run --rm -v "$PWD/fluentbit:/data" cr.fluentbit.io/fluent/fluent-bit \
  /fluent-bit/bin/fluent-bit -c /data/fluent-bit.yaml
```

Of 8 events, 3 are kept; `user` is removed and `service` added:

```json
{"date":1791736642.79,"ts":"…10:01:11Z","event":"declined","chat_id":"c-2210","reason":"no_supported_source","service":"tap-agent"}
{"date":1791736642.79,"ts":"…10:02:35Z","event":"error","chat_id":"c-3307","error":"elasticsearch timeout","latency_ms":5000,"service":"tap-agent"}
{"date":1791736642.79,"ts":"…10:03:00Z","event":"handoff","chat_id":"c-4120","reason":"low_confidence","service":"tap-agent"}
```

- Same Input → Filter → Output pipeline as Fluentd, in YAML.
- Privacy by design: log IDs and metadata, never raw message text; redact what is left in the pipeline.
- Replace `stdout` with `kafka` or `elasticsearch` to feed the next lessons (Kafka, search, observability).
